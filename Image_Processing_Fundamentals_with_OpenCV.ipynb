```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
filename = next(iter(uploaded))

img = cv2.imread(filename)

if img is None:
    raise ValueError("Görüntü okunamadı.")

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title("Orijinal ")
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.imshow(gray, cmap="gray")
plt.title("Gri Tonlamali Goruntu")
plt.axis("off")
plt.show()
```

```python
blur = cv2.GaussianBlur(gray, (5, 5), 0)

plt.imshow(blur, cmap="gray")
plt.title("Gaussian Blur")
plt.axis("off")
plt.show()
```

```python
edges = cv2.Canny(blur, 100, 200)

plt.imshow(edges, cmap="gray")
plt.title("Canny Kenar Tespiti")
plt.axis("off")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].imshow(img_rgb)
axes[0, 0].set_title("Orijinal")

axes[0, 1].imshow(gray, cmap="gray")
axes[0, 1].set_title("Gri Tonlama")

axes[1, 0].imshow(blur, cmap="gray")
axes[1, 0].set_title("Gaussian Blur")

axes[1, 1].imshow(edges, cmap="gray")
axes[1, 1].set_title("Canny Kenar Tespiti")

for ax in axes.flat:
    ax.axis("off")

plt.tight_layout()
plt.show()
```
